```python
import findspark
findspark.init()
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Socket code').master('local[3]')\
.config('spark.driver.host','127.0.0.1')\
.config('spark.ui.port',0).getOrCreate()
```

```python
spark
```

```python
df = spark.readStream.format('socket')\
.option('host','localhost')\
.option('port',1234).load()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col,split,explode
```

```python
data = df.select(explode(split(col('value'),' ')).alias('words'))
```

```python
data.printSchema()
```

```python
res = data.groupby('words').count()
```

```python
res.writeStream.format('console').outputMode('update').start()
```
