```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('file code').master('local[3]')\
.config('spark.driver.host','127.0.0.1')\
.config('spark.ui.port',0).getOrCreate()
```

```python
spark
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
sch = StructType([\
                 StructField("timestamp",StringType(),True),\
                 StructField("value",IntegerType(),True)])
```

```python
df = spark.readStream.format('json').schema(sch).load(r'/home/AzureAdmin/data/rateoutput')
```

```python
df.printSchema()
```

```python
df1 = df.select(col("timestamp").cast(TimestampType()).alias("TS"),"value")
```

```python
df1.printSchema()
```

```python
df1.writeStream.format('console').trigger(processingTime= "30 seconds").start()
```
